```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# 1. Muat Tokenizer dan Model dari Hugging Face Hub
model_name = "mdhugol/indonesia-bert-sentiment-classification"
print(f"Mengunduh dan memuat model: {model_name}...\n")

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# 2. Data Uji Mandiri (Pastikan ini buatan sendiri sesuai aturan)
data_uji = [
    "Wah, fitur baru di aplikasi ini sangat membantu pekerjaan saya sehari-hari. Terima kasih tim developer!", # Ekspektasi: Positif
    "Tolong perbaiki servernya, dari kemarin sering banget down saat lagi transaksi penting. Sangat mengecewakan.", # Ekspektasi: Negatif
    "Hari ini saya berencana pergi ke kantor menggunakan kereta KRL jalur merah." # Ekspektasi: Netral
]

# 3. Proses Inference
print("=== HASIL INFERENCE ===")
for teks in data_uji:
    # Tokenisasi teks
    inputs = tokenizer(teks, return_tensors="pt", truncation=True, padding=True)
    
    # Prediksi menggunakan model (tanpa menghitung gradien agar lebih cepat)
    with torch.no_grad():
        logits = model(**inputs).logits
    
    # Ambil label dengan probabilitas tertinggi
    predicted_class_id = logits.argmax().item()
    label = model.config.id2label[predicted_class_id]
    
    print(f"Teks Input : '{teks}'")
    print(f"Prediksi   : {label}")
    print("-" * 50)
```
